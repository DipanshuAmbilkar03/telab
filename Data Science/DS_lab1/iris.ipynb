# Data Wrangling I - Iris Dataset

This notebook completes all tasks from the first experiment using the local file `Iris.csv`.

## 1. Import Required Python Libraries

We import libraries for data handling (`pandas`, `numpy`) and normalization (`MinMaxScaler`).

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
```

## 2. Locate Open Source Data and Describe It

- **Dataset Name:** Iris Dataset
- **Open Source Website (Reference):** UCI Machine Learning Repository
- **Source URL:** https://archive.ics.uci.edu/ml/datasets/iris

### Dataset Description
- The dataset contains flower measurements for 3 iris species.
- In this local CSV, the columns are:
  - `Id`: Row identifier
  - `SepalLengthCm`, `SepalWidthCm`, `PetalLengthCm`, `PetalWidthCm`: Numeric measurements
  - `Species`: Categorical class label

### How the data is imported/read
- We use `pd.read_csv('Iris.csv')` to load the dataset from the local DS_lab1 folder.
- The file already has headers, so pandas automatically reads column names.

```python
file_path = 'Iris.csv'
df = pd.read_csv(file_path)

print('Dataset loaded from local file successfully.')
print('Shape of dataset (rows, columns):', df.shape)
print('Column names:', list(df.columns))
df.head()
```

## 3. Load Dataset into Pandas DataFrame

The dataset is loaded into the pandas DataFrame named `df`.

## 4. Data Preprocessing

Required operations in this step:
- Check missing values using `isnull()`
- Get initial statistics using `describe()`
- Provide variable descriptions and types
- Check data dimensions

```python
print('DataFrame dimensions:', df.shape)

print('\nMissing values per column:')
missing_values = df.isnull().sum()
print(missing_values)
print('Total missing values:', int(missing_values.sum()))

print('\nInitial statistics (numeric columns):')
display(df.describe())

print('Current data types:')
display(df.dtypes.to_frame(name='dtype'))

variable_descriptions = pd.DataFrame({
    'Variable': ['Id', 'SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm', 'Species'],
    'Description': [
        'Row identifier',
        'Sepal length in centimeters',
        'Sepal width in centimeters',
        'Petal length in centimeters',
        'Petal width in centimeters',
        'Iris flower species label'
    ],
    'Type': [
        'Integer',
        'Numeric (continuous)',
        'Numeric (continuous)',
        'Numeric (continuous)',
        'Numeric (continuous)',
        'Categorical (object)'
    ]
})

print('Variable descriptions:')
display(variable_descriptions)
```

## 5. Data Formatting and Data Normalization

This step includes:
- Summarizing variable types by checking data types
- Applying type conversions where required
- Normalizing numeric variables

```python
print('Data types before conversion:')
display(df.dtypes.to_frame(name='dtype'))

# Convert Species from object to category for correct semantic type
df['Species'] = df['Species'].astype('category')

print('\nData types after conversion:')
display(df.dtypes.to_frame(name='dtype'))

dtype_summary = (
    df.dtypes.astype(str)
      .value_counts()
      .rename_axis('dtype')
      .reset_index(name='count')
)
print('\nSummary of variable types:')
display(dtype_summary)

# Normalize only measurement columns; keep Id unchanged
measurement_cols = ['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm']
scaler = MinMaxScaler()
df_normalized = df.copy()
df_normalized[measurement_cols] = scaler.fit_transform(df[measurement_cols])

print('First 5 rows after Min-Max normalization (measurement columns):')
display(df_normalized.head())
```

## 6. Convert Categorical Variables into Quantitative Variables

We convert the categorical `Species` column to numeric form using one-hot encoding.

```python
df_encoded = pd.get_dummies(df_normalized, columns=['Species'], dtype=int)

print('Shape before encoding:', df_normalized.shape)
print('Shape after encoding:', df_encoded.shape)
print('\nEncoded columns:')
print([col for col in df_encoded.columns if col.startswith('Species_')])

print('First 5 rows after categorical-to-quantitative conversion:')
display(df_encoded.head())
```
