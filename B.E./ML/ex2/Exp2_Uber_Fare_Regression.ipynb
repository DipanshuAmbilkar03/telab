## Experiment 2: Uber Fare Prediction with Linear, Ridge and Lasso Regression

**Aim:** Predict the price of an Uber ride from trip features and compare Linear, Ridge and Lasso regression.

```python
# A52 Dipanshu Ambilkar
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.metrics import r2_score, mean_squared_error
```

```python
rng = np.random.default_rng(7)
n = 500
distance = rng.uniform(1, 30, n)
print("First 5 distances:", distance[:5].round(1))
```

```python
duration = distance * rng.uniform(1.5, 3, n)
passengers = rng.integers(1, 5, n)
fare = 30 + 12 * distance + 1.5 * duration + rng.normal(0, 15, n)
print("First 5 fares:", fare[:5].round(2))
```

```python
df = pd.DataFrame({"distance": distance, "duration": duration,
                   "passengers": passengers, "fare": fare})
print(df.head())
```

```python
print("Shape before cleaning:", df.shape)
```

```python
df = df.dropna()
print("Rows after cleaning:", len(df))
```

```python
q1, q3 = df["fare"].quantile([0.25, 0.75])
iqr = q3 - q1
print("Q1:", round(q1, 2), "| Q3:", round(q3, 2), "| IQR:", round(iqr, 2))
```

```python
outliers = df[(df["fare"] < q1 - 1.5 * iqr) | (df["fare"] > q3 + 1.5 * iqr)]
print("Outliers found:", len(outliers))
df = df.drop(outliers.index)
print("Rows after removing outliers:", len(df))
```

```python
print("Correlation with fare:")
print(df.corr()["fare"].round(3))
```

```python
X = df[["distance", "duration", "passengers"]]
y = df["fare"]
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=7)
print("Train size:", len(X_train), "| Test size:", len(X_test))
```

```python
model = LinearRegression()
model.fit(X_train, y_train)
pred = model.predict(X_test)
print(f"Linear   R2 = {r2_score(y_test, pred):.4f}   RMSE = {mean_squared_error(y_test, pred) ** 0.5:.2f}")
```

```python
model = Ridge()
model.fit(X_train, y_train)
pred = model.predict(X_test)
print(f"Ridge    R2 = {r2_score(y_test, pred):.4f}   RMSE = {mean_squared_error(y_test, pred) ** 0.5:.2f}")
```

```python
model = Lasso()
model.fit(X_train, y_train)
pred = model.predict(X_test)
print(f"Lasso    R2 = {r2_score(y_test, pred):.4f}   RMSE = {mean_squared_error(y_test, pred) ** 0.5:.2f}")
```
