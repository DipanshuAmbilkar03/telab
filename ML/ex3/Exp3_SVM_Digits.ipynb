## Experiment 3: SVM Classification of Handwritten Digits

**Aim:** Classify images of handwritten digits (0-9) using a Support Vector Machine.

```python
# A52 Dipanshu Ambilkar
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
X, y = load_digits(return_X_y=True)
print("Shape:", X.shape, "| classes:", sorted(set(y)))
```

```python
print("First 8 pixel values of the first image:", X[0][:8])
print("Label of the first image:", y[0])
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=7)
print("Train size:", len(X_train), "| Test size:", len(X_test))
```

```python
model = SVC()
print("Model created:", model)
```

```python
model.fit(X_train, y_train)
print("SVM trained")
```

```python
pred = model.predict(X_test)
print("First 10 predictions:", pred[:10])
print("First 10 actual labels:", y_test[:10])
```

```python
print("Accuracy:", round(accuracy_score(y_test, pred), 4))
```

```python
print("Classification report:")
print(classification_report(y_test, pred))
```

```python
print("Confusion matrix:")
print(confusion_matrix(y_test, pred))
```
